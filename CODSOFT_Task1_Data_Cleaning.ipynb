```python
import pandas as pd

```

```python
df = pd.read_csv("superstore.csv")

print("Dataset imported successfully!")
print("Shape of dataset:", df.shape)

df.head()
```

```python
# Inspect the structure of the dataset

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
# Check for missing values

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
# Count how many columns are missing in each row

missing_per_row = df.isnull().sum(axis=1)

print("Rows with missing values:", (missing_per_row > 0).sum())
print("Rows with all values missing:", (missing_per_row == len(df.columns)).sum())

# Display rows that contain missing values
df[missing_per_row > 0].head(10)
```

```python
# Identify which columns are missing in each row

missing_summary = df.isnull().sum(axis=1)

# Show the number of missing values per row
print("Distribution of missing values per row:")
print(missing_summary.value_counts().sort_index())

# Show the columns that are missing in the first 10 affected rows
rows_with_missing = df[missing_summary > 0].head(10)

for index, row in rows_with_missing.iterrows():
    missing_columns = row[row.isnull()].index.tolist()
    print(f"\nRow index {index}:")
    print("Missing columns:", missing_columns)
```

```python
# Inspect the rows with 19 missing values

incomplete_rows = df[missing_summary == 19]

print("Number of highly incomplete rows:", len(incomplete_rows))

print("\nColumns that contain data in these rows:")
print(incomplete_rows.notnull().sum())

print("\nSample of highly incomplete rows:")
display(incomplete_rows.head(10))
```

```python
# Remove rows where 19 out of 21 columns are missing

before_cleaning = len(df)

df = df[missing_summary < 19].copy()

after_cleaning = len(df)

print("Rows before removing incomplete records:", before_cleaning)
print("Rows removed:", before_cleaning - after_cleaning)
print("Rows remaining:", after_cleaning)
```

```python
# Find rows where Postal Code is missing

missing_postal = df[df['Postal Code'].isnull()]

print("Rows with missing Postal Code:", len(missing_postal))

display(missing_postal[
    ['Row ID', 'Order ID', 'City', 'State', 'Postal Code']
])
```

```python
# Fill missing Postal Codes for Burlington, Vermont
df.loc[
    (df['City'] == 'Burlington') &
    (df['State'] == 'Vermont') &
    (df['Postal Code'].isnull()),
    'Postal Code'
] = 5401

# Convert Postal Code to a 5-digit text format
df['Postal Code'] = df['Postal Code'].apply(
    lambda x: f"{int(x):05d}" if pd.notna(x) else pd.NA
)

print("Missing Postal Codes after cleaning:", df['Postal Code'].isnull().sum())
```

```python
# Check for duplicate records

duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Convert date columns to datetime format

df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
df['Ship Date'] = pd.to_datetime(df['Ship Date'], errors='coerce')

# Convert numerical columns to numeric types

numeric_columns = ['Sales', 'Quantity', 'Discount', 'Profit']

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Display the updated data types

print("Updated data types:")
print(df.dtypes)
```

```python
# Check missing values after type conversion

print("Missing values after conversion:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())

print("\nDataset shape:", df.shape)
```

```python
# Standardize text columns

text_columns = df.select_dtypes(include=['object', 'string']).columns

for col in text_columns:
    df[col] = df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("Unnecessary spaces removed from text columns.")

# Check unique categorical values
for col in ['Ship Mode', 'Segment', 'Region', 'Category', 'Sub-Category']:
    print(f"\n{col}:")
    print(sorted(df[col].dropna().unique()))
```

```python
# Final validation of cleaned dataset

print("Final dataset shape:", df.shape)

print("\nTotal missing values:", df.isnull().sum().sum())

print("Total duplicate rows:", df.duplicated().sum())

print("\nData types:")
print(df.dtypes)

print("\nFirst 5 rows:")
display(df.head())
```

```python
# Save the cleaned dataset as a CSV file

df.to_csv("superstore_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
